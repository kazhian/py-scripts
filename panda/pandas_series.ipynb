```python
import numpy as np
import pandas as pd
```

```python
arr = np.array([1, 2, 3, 4, '5'], dtype=np.int32)
my_series = pd.Series(arr)
print(my_series)
```

```python
my_series1 = pd.Series(arr, index=['a', 'b', 'c', 'd', 'e'])
print(my_series1)
```

```python
my_series2 = pd.Series({'a': 10, 'x': 20, 'y': 30})
print(my_series2)
```

```python
my_series1 + my_series2 # This is like a database inner join operation
```

```python
# a series with 10 elements from 1 to 10 with index as alphabets from a to j
series = pd.Series(np.arange(1, 11), index=[chr(i) for i in range(97, 107)])
print(series)
# select multiple labels by passing a list (or use .loc)
print(series[['a', 'e', 'i']])
```
