# Short conceptual map (Excel analogy)

## Excel analogy

- Python list: a single unordered column of cells with no labels — just values. Can hold mixed types. No built‑in vectorized math or alignment.
- NumPy array: like a fast Excel column/array of numbers only (homogeneous dtype), stored contiguously. Great for numeric vectorized ops, but no row/column labels.
- pandas.Series: an Excel column with an index (row labels). 1D, supports labels, alignment, and many series methods. Under the hood it wraps a NumPy array + index.
- pandas.DataFrame: a full Excel sheet — 2D table with column headers and row index. Each column is a Series; different columns can have different dtypes.

### Key differences (concise)

- Dimensionality: list/Series/1D array = 1D, DataFrame = 2D.
- Labels: list/NumPy = position-only; Series/DataFrame = label-aware (index, columns).
- Homogeneity: list can be mixed; NumPy usually homogeneous; Series usually homogeneous (per Series) but may be object dtype; DataFrame columns can each have their own dtype.
- Alignment: Series/DataFrame align on labels for arithmetic; lists/NumPy align by position.
- Missing data: pandas has built-in NaN/NA handling and many methods for filling/dropping; lists/NumPy do not offer the same higher-level semantics.
- Indexing: .loc uses labels, .iloc uses integer positions (DataFrame/Series).
- Use cases: lists for small heterogeneous collections, NumPy for heavy numeric computation, Series for labeled 1D data (time series), DataFrame for tabular datasets.

```python
import numpy as np
import pandas as pd
```

```python
# list
L = [1, 'a', 3.5]         # heterogeneous, no labels
print("List:", L)
```

```python
# numpy array (homogeneous)
A = np.array([1, 2, 3])   # fast numeric ops: A + 1 -> [2,3,4]
print("Numpy Array:", A)
# Series is vectorized. You can do operations like A + 1, A * 2, etc.
print("Numpy Array + 1:", A + 1)
```

```python
# Series (1D labeled)
list = [10, 20, 30]
s = pd.Series(list, index=['row1', 'row2', 'row3'])
# like an Excel column with row labels
print(s['row2'])          # 20
print(s + 1)              # elementwise +1, preserves labels
```

```python
# DataFrame (2D labeled)
df = pd.DataFrame({
    'Sales': [100, 200],
    'Region': ['US', 'EU']
}, index=['Jan', 'Feb'])
# like an Excel sheet with headers and row labels
print(df.loc['Jan', 'Sales'])   # 100
print(df['Sales'] * 2)          # vectorized op on a column (Series)

# alignment example
s1 = pd.Series([1, 2], index=['a', 'b'])
s2 = pd.Series([10,20], index=['b', 'c'])
print(s1 + s2)   # result aligned on index: a->NaN, b->22, c->NaN
```

## When to pick which

- Use list: tiny, heterogeneous collections; simple Python logic.
- Use NumPy: numeric arrays, fast elementwise/math ops, linear algebra.
- Use Series: labeled one-dimensional data (time series, single column with index).
- Use DataFrame: tabular data with multiple columns, mixed types, indexing and group/aggregate operations (closest to Excel for data analysis).
