```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 4, 5], dtype=np.int32)
print("Numpy Array:", arr)
print("Array Shape:", arr.shape)
print("Array Data Type:", arr.dtype)
print("items form index 1 to 3:", arr[1:4])
```

```python
print(np.ones((3, 4)))
print(np.zeros((2, 5)))
```

```python
print(np.ones(10))
print(np.ones(10) / 2)
```

```python
np.arange(0, 20, 2)
```

```python
# Linear space between 0 and 1 with 5 values
np.linspace(0, 1, 5) 
```

```python
num_list = [1, 2, 3, 4, 5]
# Do it in traditional way, list comprehension and map function
print([x * 2 for x in num_list])
print(list(map(lambda x: x * 2, num_list)))

# Using numpy array for vectorized operations
num_array = np.array(num_list)
print(num_array * 2)
print(num_array > 2)
```

```python
arr = np.array([1, 2, 3, 4, 5], dtype=np.int32)
print(arr[True, False, True, False, True])
print(arr[arr > 2])
print(arr[arr % 2 == 0])
```
